## 1. The machine repeats what you say

Type this and run it:

````python
print("hello")
````

The machine prints `hello`. This is the whole transaction: you write an instruction, the machine carries it out, exactly, and nothing more. It does not know what you meant. It knows what you wrote. The entire craft of programming lives in that gap, and this book is partly about learning to feel the gap before the machine shows it to you.

## 2. The machine does arithmetic

````python
print(3 + 4)
print(3 * 4)
print(2 ** 10)
````

The last line is 2 to the 10th power. Run it. Then change it to `2 ** 100` and run it again. That number has thirty-one digits. The machine did not slow down. Hold on to this: the machine is not smart, but it is *fast*, and almost everything in this book comes from asking what fast-but-not-smart is good for.

## 3. The machine remembers what you name

````python
steps = 0
steps = steps + 1
steps = steps + 1
print(steps)
````

A *variable* is a name attached to a value. The line `steps = steps + 1` looks wrong if you read `=` as it is read in mathematics — no number equals itself plus one. In Python, `=` is not a statement of fact. It is an instruction: *compute the thing on the right, then attach the name on the left to it.* Read it as "steps becomes steps plus one" and it is sensible.

This little program is a counter. In Chapter 1 we will count something worth counting.

## 4. The machine tells you when you've lost it

Run this. It is broken on purpose.

````python
print("the answer is " + 42)
````

Python stops and prints several lines of complaint ending in something like:

```
TypeError: can only concatenate str (not "int") to str
```

Read the last line. It is not scolding you; it is reporting, precisely, what it could not do: glue a piece of text to a number, because those are different kinds of things. An error message is information. Every programmer who has ever lived reads dozens of them a day. The ones who get good are the ones who read them rather than flinch.

Fix the program by making the number into text:

````python
print("the answer is " + str(42))
````

## 5. The machine is fast

One more, and this one needs a construct we will not explain until Chapter 1 — run it anyway:

````python
total = 0
for n in range(1, 1000001):
    total = total + n
print(total)
````

This adds every whole number from 1 to one million. Run it and notice two things. First, the answer appears before your finger leaves the key. A million additions cost the machine nothing you can perceive. Second — check the answer. Gauss, as a schoolboy, is said to have computed this sum (to one hundred, but the trick scales) in seconds, by *thinking* instead of adding: pair 1 with 1,000,000, pair 2 with 999,999, and so on; 500,000 pairs, each summing to 1,000,001.

So there are two ways to be fast: do the dumb thing very quickly, or find a way to not do most of the work at all. The machine gives you the first for free. This book is about the second. An *algorithm* is, more than anything else, a way of not doing most of the work — and the strange, wonderful fact is that the not-doing can be written down, studied, and proved correct, like any other piece of mathematics.

That is where Chapter 1 begins.
